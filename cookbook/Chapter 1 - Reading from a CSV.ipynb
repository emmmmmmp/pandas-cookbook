```python
#chapter1 of pandas as in polars

import polars as pl
import matplotlib.pyplot as plt

plt.style.use("ggplot")
plt.rcParams["figure.figsize"] = (15, 5)

# Read the CSV without correcting the separator
broken_df = pl.read_csv(
    "./data/bikes.csv",
    encoding="latin1"
)

print(broken_df.head(3))

# Correctly load and parse the data
fixed_df = (
    pl.read_csv(
        "./data/bikes.csv",
        separator=";",
        encoding="latin1"
    )
    .with_columns(
        pl.col("Date").str.to_date("%d/%m/%Y")
    )
)

print(fixed_df.head(3))

# Select the Berri 1 column
print(fixed_df.select("Berri 1"))

# Plot Berri 1 cycling traffic
plt.figure(figsize=(15, 5))
plt.plot(fixed_df["Date"], fixed_df["Berri 1"])
plt.xlabel("Date")
plt.ylabel("Number of cyclists")
plt.title("Berri 1 cycling traffic")
plt.show()

# Plot all cycling routes
plt.figure(figsize=(15, 10))

for column in fixed_df.columns:
    if column != "Date":
        plt.plot(
            fixed_df["Date"],
            fixed_df[column],
            label=column
        )

plt.xlabel("Date")
plt.ylabel("Number of cyclists")
plt.title("Cycling traffic by route")
plt.legend()
plt.show()

# Reproduce the final example
df = (
    pl.read_csv(
        "./data/bikes.csv",
        separator=";",
        encoding="latin1"
    )
    .with_columns(
        pl.col("Date").str.to_date("%d/%m/%Y")
    )
)

plt.figure(figsize=(15, 5))
plt.plot(df["Date"], df["Berri 1"])
plt.show()
```
